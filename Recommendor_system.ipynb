```python
# Import libraries
import pandas as pd
import numpy as np
!pip install scikit-surprise
from surprise import SVD, Dataset, Reader
from surprise.model_selection import train_test_split
from surprise.accuracy import rmse

# Load MovieLens Dataset
!wget -nc https://files.grouplens.org/datasets/movielens/ml-latest-small.zip
!unzip -n ml-latest-small.zip

# Load data
ratings = pd.read_csv('ml-latest-small/ratings.csv')
movies = pd.read_csv('ml-latest-small/movies.csv')

# Prepare Surprise dataset
reader = Reader(rating_scale=(0.5, 5.0))
data = Dataset.load_from_df(ratings[['userId', 'movieId', 'rating']], reader)

# Train SVD model
trainset = data.build_full_trainset()
algo = SVD()  # Singular Value Decomposition
algo.fit(trainset)

# Select user ID
user_id = 1  # Replace with any user ID in the dataset

# Get the movies this user has already rated
user_movies = ratings[ratings['userId'] == user_id][['movieId', 'rating']]

# Get all movie IDs
all_movie_ids = movies['movieId'].unique()

# Find movies the user hasn't rated
unseen_movie_ids = [movie for movie in all_movie_ids if movie not in user_movies['movieId'].values]

# Predict ratings for unseen movies
predictions = [algo.predict(user_id, movie_id) for movie_id in unseen_movie_ids]
predicted_ratings = [(pred.iid, pred.est) for pred in predictions]

# Sort movies by predicted rating
recommended_movies = sorted(predicted_ratings, key=lambda x: x[1], reverse=True)[:10]

# Get recommended movie titles
recommended_movie_ids = [movie[0] for movie in recommended_movies]
recommended_movie_titles = movies[movies['movieId'].isin(recommended_movie_ids)]['title']

print(f"Recommended Movies for User {user_id}:")
print(recommended_movie_titles.values)
```
